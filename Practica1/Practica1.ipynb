```python
#Ejercicio1

def is_integer(text: str) ->bool:
    """
    Devuelve True si la cadena es un numero entero, sino, devuelve False.

    Par:
        string text: cadena a evaluar
    Devuelve:
        bool: True si es un numero entero, False si no lo es
    """
    try:
        int(text)
        return True
    except ValueError:
        return False

def seconds_to_dhms(total_seconds: int) ->str:
    """
    Convierte los segundos a días, horas, minutos y segundos

    Par:
        int total_seconds: segundos
    Devuelve:
        str: días, horas, minutos y segundos equivalentes a los segundos introducidos
    """
    if total_seconds>=0:
        dias = total_seconds//86400
        total_seconds %= 86400
        horas = total_seconds//3600
        total_seconds %= 3600
        minutos = total_seconds//60
        segundos = total_seconds%60

        return dias, horas, minutos, segundos
    else:
        print("Introduzca un entero no negativo")

def main():
    while True:
        text = input("Introduzca los segundos: ")
        if is_integer(text):
            total_seconds = int(text)
            if total_seconds >= 0:
                break
        print("Entrada no válida. Introduzca un entero no negativo.")

    days, hours, minutes, seconds = seconds_to_dhms(total_seconds)
    print(f"{total_seconds} segundos son {days} d, {hours} h, {minutes} min y {seconds} s.")

if __name__ == "__main__":
    main()
```

```python
#Ejercicio2

def is_leap_year(year: int) -> bool:
    """
    Devuelve True si el año es bisiesto y False si no lo es

    Par:
        int year: año a evaluar
    Devuelve:
        bool: True si es bisiesto y False si no
    """
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

def leap_year_between(start: int, end: int) -> list[int]:
    """
    Devuelve la lista de los años bisiestos entre start y end

    Par:
        int start: año de inicio de la lista
        int end: año de fin de la lista
    Devuelve:
        list[int]: lista de los años bisiestos entre start y end
    """
    bis = []

    for year in range(start, end+1):

        if is_leap_year(year):
            bis.append(year)
    return bis

def explain_leap_year(year: int) -> str:
    """
        Indica si el año es bisiesto y por qué
        Par:
            int year: año a evaluar
        Devuelve:
            str: cadena con la expliación de por qué es o no año bisiesto
    """
    if year % 400 == 0:
        return f"{year} es bisiesto: es divisible entre 400."
    elif year % 100 == 0:
        return f"{year} no es bisiesto: es divisible entre 100 pero no entre 400."
    elif year % 4 == 0:
        return f"{year} es bisiesto: es divisible entre 4 y no entre 100."
    else:
        return f"{year} no es bisiesto: no es divisible entre 4."


def main():
    while True:
        year = int(input("Año: "))
        if year < 1582:
            print("El calendario gregoriano se estableció en 1582")
        else:
            print(explain_leap_year(year))
            break

    print("Ejemplo de uso de leap_years_between entre 1896 y 1912")
    print(leap_year_between(1896, 1912))

if __name__ == "__main__":
    main()
    
```

```python
#Ejercicio3

def process_text(text: str = "This is a default text with Python and amazing words.",
                  words_replace: list[str] = ["Python", "amazing"] ) -> tuple[str, int]:
    """
        Busca en la cadena de texto las palabras de la lista y las remplaza por asteriscos

        Par:
             str text: cadena de texto a evaluar, valor predeterminado por defecto
             lista words_replace: lista de palabras a buscar y reemplazar
        Devuelve:
             string processed_words: la cadena de texto en minúsculas con las palanras coincidentes reemplazadas por asteriscos
             int count: el número de palabras que han sido reemplazadas
    """
    text = text.lower()
    cleaned_text = text.strip().lower()

    rem_lower = []
    for w in words_replace:
        rem_lower.append(w.lower())

    
    words = cleaned_text.split()
    processed_words = []
    count = 0

    for w in words:
        if w in rem_lower:
            processed_words.append("*" * len(w))
            count += 1
        else:
            processed_words.append(w)

    processed_text = " ".join(processed_words)

    return processed_text, count

```

```python
# Ejercicio 4

def shift_char(char: str, shift: int)->str:
    """
    Desplaza un carácter alfabético un número de posiciones.

    Si el carácter es una letra minúscula o mayúscula, se desplaza
    manteniendo su capitalización. Si no es una letra inglesa,
    se devuelve sin modificar.

    El desplazamiento puede ser negativo o mayor que 26.

    Par:
        char (str): Carácter que se desea desplazar.
        shift (int): Número de posiciones que se desplaza el carácter.

    Devuelve:
        str: Carácter desplazado o carácter original si no es una letra.

    """

    if 'a'<=char<='z':
        char = chr((ord(char) - ord('a') + shift) % 26 + ord('a'))
    elif  'A'<=char<='Z':
        char = chr((ord(char) - ord('A') + shift) % 26 + ord('A'))
    return char


def encrypt(text: str, shift: int)->str:
    """
    Cifra un texto utilizando el cifrado César.

    Desplaza cada letra el número de posiciones indicado.
    Mantiene las mayúsculas, las minúsculas y los caracteres
    que no pertenecen al alfabeto inglés sin cambios.

    Par:
        text (str): Texto original que se desea cifrar.
        shift (int): Desplazamiento del cifrado.

    Devuelve:
        str: Texto cifrado.
    """
    en_text = ""
    for char in text:
        en_text += shift_char(char, shift)

    return en_text

def decrypt(text: str, shift: int)->str:
    """
    Descifra un texto cifrado mediante el cifrado César.

    Reutiliza la función encrypt() aplicando el desplazamiento
    contrario al utilizado para cifrar.

    Par:
        text (str): Texto cifrado que se desea descifrar.
        shift (int): Desplazamiento utilizado al cifrar.

    Devuelve:
        str: Texto descifrado.
    """
    decr_text = ""
    shift = -shift
    decr_text = encrypt(text, shift)
    return decr_text

def brute_force(cipher_text: str)->list[str]:
    """
    Genera los 26 posibles descifrados de un texto cifrado.

    Prueba todos los desplazamientos desde 0 hasta 25,
    utilizando la función decrypt().

    Par:
        cipher_text (str): Texto cifrado que se desea analizar.

    Devuelve:
        list[str]: Lista con los 26 posibles textos descifrados,
        ordenados según el desplazamiento de 0 a 25.
    """
    lista = []
    for shift in range(26):
        decr_text = encrypt(cipher_text, shift) 
        lista.append(decr_text)
    return lista



def main():
    text = input("Introduzca el texto a encriptar: ")
    shift = int(input("Introduzca el número de posiciones a desplazar: "))
    
    cipher_text = encrypt(text, shift)
    print(f"Texto encriptado: {cipher_text}")
    
    decrypted_text = decrypt(cipher_text, shift)
    print(f"Texto desencriptado con Clave Cesar: {decrypted_text}")

    print("Prueba Fuerza Bruta: ")
    text2 = input("Introduzca el texto a encriptar: ")
    shift2 = int(input("Introduzca el número de posiciones a desplazar: "))
    
    cipher_text2 = encrypt(text2, shift2)
    print(f"Texto encriptado: {cipher_text2}")

    print("Ataque por Fuerza Bruta: ")
    
    
    
    posibilidades = brute_force(cipher_text2)

    print("Probando los 26 desplazamientos posibles:")
    for shift in range(len(posibilidades)):
        opcion = posibilidades[shift]
        print(f"Shift {shift:2d}: {opcion}")
    
    
if __name__ == "__main__":
    main()

```

```python
#Ejercicio 5

def normalize(text:str)->str:
    """
    Normaliza una cadena de texto.

    Convierte el texto a minúsculas, elimina las tildes de las vocales
    y la diéresis de la u,  elimina los espacios y otros caracteres
    que no sean letras ni números.

    Par:
        text (str): Cadena de texto que se desea normalizar.

    Devuelve:
        str: Texto normalizado.
    """
    low_text = text.lower() #Convertimos el texto en minúsculas
    low_text = low_text.replace("á", "a") #Reemplazo de las tildes y diéresis
    low_text = low_text.replace("é", "e")
    low_text = low_text.replace("í", "i")
    low_text = low_text.replace("ó", "o")
    low_text = low_text.replace("ú", "u")
    low_text = low_text.replace("ü", "u")

    norm_text = ""

    for char in low_text:
        if char.isalnum():
            norm_text += char

    return norm_text

def is_palindrome(text: str)->bool:
    """
    Comprueba si una cadena de texto es un palíndromo.

    Normaliza el texto y lo compara con su versión invertida.

    Par:
        text (str): Cadena de texto que se desea comprobar.

    Devuelve:
        bool: True si el texto es un palíndromo, False en caso contrario.
    """
    norm_text = normalize(text)
    prueba = norm_text[::-1]
    if prueba == norm_text:
        return True
    else:
        return False

def find_palindromes(sentences: list[str])->list[str]:
    """
    Busca los palíndromos dentro de una lista de frases.

    Par:
        sentences (list[str]): Lista de frases que se desean comprobar.

    Devuelve:
        list[str]: Lista de frases que son palíndromos.
    """
    pal = []
    for word in sentences:
        if is_palindrome(word):
            pal.append(word)
    return pal

def palindrome_words(text: str)->list[str]:
    """
    Busca las palabras palíndromas de al menos tres caracteres
    dentro de una frase.

    Separa la frase en palabras y conserva la escritura original
    de cada palabra encontrada.

    Par:
        text (str): Frase sin signos de puntuación.

    Devuelve:
        list[str]: Lista de palabras palíndromas de al menos
        tres caracteres.
    """
    split_text= text.split()
    pal_words = []
    for word in split_text:
        if len(word) >= 3 and is_palindrome(word):
            pal_words.append(word)

    return pal_words



def main():
    text = input("Introduzca el texto para encontrar los palíndromos: ")
    pal_words = palindrome_words(text)
    print("Cadena con los palíndromos:")
    for word in pal_words:
        print(f"\n{word}")

if __name__ == "__main__":
    main()

```

```python
#Ejercicio 6

def is_prime(n: int)->bool:
    """
    Comprueba si un número entero n es primo.

    Un número primo es un entero mayor que 1 que solo es divisible
    por 1 y por sí mismo.

    Args:
        n (int): Número a evaluar.

    Returns:
        bool: True si n es primo, False en caso contrario.
    """
    if n < 2:
        return False
    else:
        for i in range(2,n):
            if n % i == 0:
                return False
    return True

def prime_list(limit: int)->list[int]:
    """
    Genera una lista de números primos hasta un límite superior (incluido).

    Args:
        limit (int): Límite máximo para buscar números primos.

    Returns:
        list[int]: Lista con los números primos encontrados. Si limit < 2,
                   devuelve una lista vacía.
    """
    primes =[]
    for num in range(limit+1):
        if is_prime(num):
            primes.append(num)
    return primes

def check_palindrome(primes: list[int])->list[int]:
    """
    Filtra y devuelve los números primos que son palíndromos (capicúa).

    Args:
        primes (list[int]): Lista de números primos a evaluar.

    Returns:
        list[int]: Lista con los números primos palíndromos.
    """
    palindromes = []
    for n in primes:
        if n>9:
            if str(n) == str(n)[::-1]:
                palindromes.append(n)
    return palindromes

def categorize_prime(prime: int)->str:
    """
    Clasifica un número primo según su valor en:
    - "pequeño": menor que 10.
    - "mediano": entre 10 y 99.
    - "grande": 100 o más.

    Args:
        prime (int): Número primo a clasificar.

    Returns:
        str: Categoría del número primo ("pequeño", "mediano" o "grande").
    """
    if prime < 10:
        return "pequeño"
    elif prime < 100:
        return "mediano"
    else:
        return "grande" 

def main ():
    print("Comprobación is_prime: ")
    num = int(input("Introduzca el número:"))
    if is_prime(num):
        print(f"El número {num} es primo")
        print("Comprobación categorize_prime: ")
        cad = ""
        cad = categorize_prime(num)
        print(f"El numero es {cad}")
    else:
        print(f"El número {num} no es primo")

    

    print("Comprobación prime_list: ")
    lim = int(input("Introduzca el límite:"))
    list_primes = []
    list_primes = prime_list(lim)
    print("Lista de enteros: \n")
    for num in list_primes:
        print(num)

    print("Comprobación check_palindrome: ")
    pal = []
    pal = check_palindrome(list_primes)
    print("Lista de palíndromos: \n")
    for num in pal:
        print(num)




if __name__ == "__main__":
    main()
    

```

```python
#Ejercicio 7

def remove_duplicates(numbers:list[int])->list[int]:
    """Elimina los elementos duplicados de una lista manteniendo el orden original.

    Par: numbers: Lista de números enteros de entrada.
    Devuelve: Una nueva lista con los elementos sin duplicados.
    """
    nd_numbers = []
    for i in numbers:
        ya_existe = False
        for j in nd_numbers:
            if i==j:
                ya_existe = True
                break
        if not ya_existe:
            nd_numbers.append(i)
    return nd_numbers
def rotate_left(numbers: list[int], k: int)->list[int]:
    """Rota los elementos de una lista k posiciones hacia la izquierda.

    Par: numbers: Lista de números enteros a rotar.
    Par: k: Número de posiciones a desplazar a la izquierda.
    Devuelve: La lista resultante tras aplicar la rotación.
    """
    rotated = []
    aux = []
    if k == 0 or len(numbers)==0:
        rotated. append(numbers)
        return rotated
    k_rot = k % len(numbers)
    #rotación de la lista
    rotated = numbers[k_rot:] + numbers[:k_rot]
    return rotated

def split_even_odd(numbers: list[int])->tuple[list[int], list[int]]:
    """Separa los números de una lista en dos colecciones: pares e impares.

    Par: numbers: Lista de números enteros a evaluar.
    Devuelve: Una tupla con dos listas (lista_pares, lista_impares).
    """
    pares = []
    impares = []
    for i in numbers:
        if i%2 == 0:
            pares.append(i)
        else:
            impares.append(i)
    return pares, impares

def merge_sorted(a: list[int], b: list[int])->list[int]:
    """Combina dos listas previamente ordenadas en una sola lista ordenada.

    Par: a: Primera lista ordenada de números enteros.
    Par: b: Segunda lista ordenada de números enteros.
    Devuelve: Una nueva lista combinada y ordenada de menor a mayor.
    """
    ordenada = []
    i = 0
    j = 0
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            ordenada.append(a[i])  
            i += 1                
        else:
            ordenada.append(b[j])  
            j += 1                 

    # Añadimos lo que haya quedado sobrante en 'a' o 'b'
    ordenada.extend(a[i:])
    ordenada.extend(b[j:])

    return ordenada

def second_largest(numbers: list[int])->int|None:
    """Encuentra el segundo valor más grande de una lista en una sola pasada.

    Par: numbers: Lista de números enteros.
    Devuelve: El segundo valor más grande distinto del máximo, o None si no existe.
    """
    max1 = None  # El número más grande encontrado
    max2 = None  # El segundo más grande (estrictamente menor que max1)

    for i in numbers:
        # 1. Si encontramos un número estrictamente mayor que el máximo actual
        if max1 is None or i > max1:
            max2 = max1
            max1 = i
        # 2. Si el número no supera a max1, pero supera a max2 y no es igual a max1
        elif i != max1 and (max2 is None or i > max2):
            max2 = i
    return max2

def main():
    lista = [2, 3, 4, 4, 5, 7, 8, 9, 2, 11]
    nd_lista = []
    nd_lista += remove_duplicates(lista)
    print("Lista sin duplicados")
    for i in nd_lista:
        print(f"{i} ")
        
    print("Rotación lista")
    rot = []
    k = int(input("Posiciones a rotar: "))
    rot += rotate_left(lista, k)
    for i in rot:
            print(f"{i} ")

        
    pares, impares = split_even_odd(lista)
    print("Lista pares: ")
    for i in pares:
            print(f"{i} ")

    print("Lista impares: ")
    for i in impares:
            print(f"{i} ")

    lista_a = [1, 4, 8, 10, 15]
    lista_b = [0, 3, 9, 11, 30]
    lista_ordenada = merge_sorted(lista_a, lista_b)
    print("Lista ordenada: ")
    for i in lista_ordenada:
        print(f"{i} ")
    lista_mayor = [5, 8, 2, 25, 3]
    lista_mayor2 = [2,2,2]
    second_max = second_largest(lista_mayor)
    print(f"Segundo mayor: {second_max}")
    second_max = second_largest(lista_mayor2)
    print(f"Segundo mayor: {second_max}")
    



if __name__ == "__main__":
    main()
    


```

```python
#Ejercicio 8
def mostrar_menu()->str:
    """Muestra el menú de opciones en consola y lee la elección del usuario.

    Devuelve: La opción seleccionada como cadena de texto.
    """
    print("\n--- MENÚ DE OPCIONES ---")
    print("1. Calcular la suma de todos los números")
    print("2. Calcular la suma de positivos y negativos por separado")
    print("3. Encontrar el número máximo y el mínimo")
    print("4. Comprobar si todos los números son únicos")
    print("5. Comparar cantidad de números positivos y negativos")
    print("6. Salir")
    return input("Selecciona una opción (1-6): ").strip()

def pedir_lista() -> list[int]:
    """Lee números enteros uno a uno mediante un bucle hasta que se indique parar.

    Devuelve: Lista de enteros ingresados por el usuario.
    """
    numbers = []
    print("Introduce los números de la lista de uno en uno.")
    print("Escribe 'fin' o pulsa Enter sin escribir nada para terminar.")

    while True:
        entrada = input("Introduce un número entero: ").strip() #Elimina espacios en blanco al inicio o al final
        
        # Condición de salida del bucle
        if entrada.lower() == "fin" or entrada == "":
            break
            
        try:
            numero = int(entrada)
            numbers.append(numero)
        except ValueError:
            print("Error: Debes escribir un número entero válido o 'fin' para salir.")

    return numbers
def suma(numbers: list[int])->int:
    """Calcula la suma total de los elementos de una lista.

    Par: numbers: Lista de enteros.
    Devuelve: La suma de todos los elementos (0 si la lista está vacía).
    """
    sum = 0
    for i in numbers:
        sum += i
    return sum

def sum_signos(numbers: list[int])->tuple[int, int]:
    """Suma los números positivos y negativos por separado. El 0 no cuenta.

    Par: numbers: Lista de enteros.
    Devuelve: Una tupla con (suma_positivos, suma_negativos). 0 si está vacía.
    """
    sum_positivos = 0
    sum_negativos = 0
    for i in numbers:
        if i >0:
            sum_positivos += i
        elif i<0:
            sum_negativos +=i
    return sum_positivos, sum_negativos

def max_min(numbers: list[int])->tuple[int, int]:
    """Encuentra el valor máximo y mínimo de la lista.

    Par: numbers: Lista de enteros.
    Devuelve: Una tupla (máximo, mínimo) o (None, None) si la lista está vacía.
    """
    if not numbers:
        return None, None
    mayor = max(numbers)
    menor = min(numbers)
    return mayor, menor

def unicos(numbers: list[int])->bool:
    """Comprueba si todos los elementos de la lista son únicos.

    Par: numbers: Lista de enteros.
    Devuelve: True si todos los elementos son únicos o si está vacía; False en caso contrario.
    """
    repetidos = []
    for i in numbers:
        if i in repetidos:
            return False
        else:
            repetidos.append(i)
    return True

def positivos_negativos(numbers: list[int])->str:
    """Compara la cantidad de números positivos y negativos.

    Par: numbers: Lista de enteros.
    Devuelve: Mensaje indicando si hay más positivos, negativos o si hay empate.
    """
    cont_pos = 0
    cont_neg = 0
    for i in numbers:
        if i > 0:
            cont_pos +=1
        elif i<0:
            cont_neg +=1
    if cont_pos > cont_neg:
        return ("La lista contiene más números positivos")
    elif cont_neg > cont_pos:
        return ("La lista contiene más números positivos")
    else:
        return ("La lista contiene la misma cantidad de  números positivos y negativos")


def main():
    lista = []
    lista += pedir_lista()
    opcion = ""
    while opcion != "6":
        opcion = mostrar_menu()
        match opcion:
            case "1":
                suma_lista = suma(lista)
                print(f"La suma de todos los elemntos de la lista es: {suma_lista}")

            case "2":
                sum_pos, sum_neg = sum_signos(lista)
                print(f"Suma de positivos: {sum_pos}") 
                print(f"Suma de negativos: {sum_neg}")

            case "3":
                maximo, minimo = max_min(lista)
                if maximo is None or minimo is None:
                    print("La lista está vacía, no tiene ni máximo ni mínimo")
                else:
                    print(f"Máximo de la lista: {maximo}. Mínimo: {minimo}")

            case "4":
                if unicos(lista):
                    print("Todos los elemntos de la lista son únicos")
                else:
                    print("Hay elementos repetidos en la lista")

            case "5":
                print(f"{positivos_negativos(lista)}")

            case "6":
                print("Fin del programa")

            case _:
                print("Opción no válida. Seleccione una opción(1-6)")


if __name__ == "__main__":
    main()


```

```python
#Ejercico 9
def format_monomial(coef: int, exp: int)->str:
    """
    Formatea un único término (monomio) según su coeficiente y exponente.

    Par:
        coef (int): Coeficiente numérico del monomio.
        exp (int): Exponente al que está elevada la variable.

    Devuelve:
        str: Representación en cadena del monomio (ej. '+2x**3', '-x', '+5').
             Devuelve una cadena vacía si el coeficiente es 0.
    """
    if coef == 0:
        return ""
    #Condiciones coeficiente
    if coef == 1:
        monomio = "+"
    elif coef == -1:
        monomio = "-"
    elif coef > 0:
        monomio = f"+{coef}"
    else:
        monomio = str(coef)

    #Condicones exponente
    if exp == 1:
        monomio += "x"
    elif exp == 0:
        pass
    else:
        monomio += f"x**{exp}"

    return monomio
def format_polynomial(coefs: list[int])->str:
    """
    Construye la representación legible de un polinomio a partir de sus coeficientes.

    Par:
        coefs (list[int]): Lista de coeficientes ordenados de mayor a menor grado.

    Devuelve:
        str: Cadena formateada del polinomio (ej. '3x**2 - 2x + 1'). 
             Devuelve '0' si el polinomio no tiene términos no nulos.
    """
    exp = len(coefs) -1
    polinomio = ""
    for coef in coefs:
        monomio = format_monomial(coef, exp)
        exp -= 1
        if monomio == "":
            continue
        if polinomio ==  "":
            if monomio.startswith("+"): #Si el primer término es positivos, quitamos el +
                monomio = monomio[1:]
            polinomio = monomio
        else:
            if monomio.startswith("+"):
                polinomio += " + " + monomio[1:]
            elif monomio.startswith("-"):
                polinomio += " - " + monomio[1:]
            else:
                polinomio += " + " + monomio

    if polinomio == "":
        return "0"

    return polinomio
def evaluate(coefs: list[int], x: int)->int:
    """
    Evalúa el polinomio en un valor determinado de x mediante sustitución directa.

    Par:
        coefs (list[int]): Lista de coeficientes ordenados de mayor a menor grado.
        x (int): Valor numérico en el que se evaluar el polinomio.

    Devuelve:
        int: Resultado numérico de calcular P(x). Devuelve 0 si la lista está vacía.
    """
    if not coefs:
        return 0
    total = 0
    n = len(coefs)
    for i, coef in enumerate(coefs):
        exp = n - 1 - i
        total += coef*(x**exp)
    return total

def derivate(coefs: list[int])->list[int]:
    """
    Calcula la primera derivada del polinomio utilizando la regla de la potencia.

    Par:
        coefs (list[int]): Lista de coeficientes del polinomio original (de mayor a menor grado).

    Devuelve:
        list[int]: Lista de coeficientes del polinomio derivado (de mayor a menor grado).
                   Devuelve una lista vacía si el polinomio es constante o está vacío.
    """
    n = len(coefs)
    if len(coefs) <= 1:
            return []
    derivada = []
    for i in range(n-1):
        exp = n -1 - i
        derivada.append(coefs[i]*exp)
    return derivada
        
def main():
    print("Función monomial_format")
    coeficiente = int(input("Introduzca el coeficiente del monomio "))
    exponente = int(input("Introduzca el exponente del monomio "))
    mon = format_monomial(coeficiente, exponente)
    print(f"Monomio: {mon}")

    coefs = []
    print("Introduce los coeficientes del polinomio, para salir pulsa Enter sin escribir nada: ")
    while True:
        num = input("Introduce un coeficiente: ")
        if num.strip() == "":
            break
        coefs.append(int(num))
    pol = format_polynomial(coefs)
    print(f"Polinomio: {pol}")

    x = int(input("Introduce el valor a evaluar en el polinomio: "))
    res = evaluate(coefs, x)
    print(f"Resultado: {res}")

    der = derivate(coefs)
    print("Polinomio derviado: ")
    print(f"Coeficientes derivados: {der}")
    print(f"Polinomio derivado: {format_polynomial(der)}")


if __name__ == "__main__":
    main()
```

```python
#Ejercicio 10
def mostrar_menu()->str:
    """Muestra el menú de opciones en consola y lee la elección del usuario.

    Devuelve: La opción seleccionada como cadena de texto.
    """
    print("\n--- MENÚ DE OPCIONES ---")
    print("1. Imprimir cadenas por letra inicial")
    print("2. Contar cadenas con una subcadena")
    print("3. Encontrar la cadena más larga y más corta")
    print("4. Verificar identidad de objeto")
    print("5. Verificar cadenas mayores a 10 caracteres")
    print("6. Salir")
    return input("Selecciona una opción (1-6): ").strip()

def pedir_cadenas() -> list[str]:
    """Lee cadenas de texto de una en una desde el teclado hasta que se indique parar.

    Devuelve: Lista de cadenas ingresadas por el usuario.
    """
    cadenas = []
    print("Introduce las cadenas de texto de una en una.")
    print("Pulsa Enter sin escribir nada (o escribe 'fin') para terminar.\n")

    while True:
        texto = input("Introduce un texto: ").strip() #Elimina espacios en blanco al inicio o al final

        # Condición de parada (presionar Enter o escribir 'fin')
        if texto.lower() == "fin" or texto == "":
            break

        cadenas.append(texto)

    return cadenas



def imprimir_por_inicial(cadenas: list[str], letra: str)->None:
    """Imprime por pantalla las cadenas de la lista que comienzan con la letra especificada.

    Par:
        cadenas (list[str]): Lista de cadenas de texto a evaluar.
        letra (str): Carácter o prefijo con el que deben iniciar las cadenas.
    """
    for cad in cadenas:
        if cad.startswith(letra):
            print(f"{cad}")

def contar_subcadenas(cadenas: list[str], subcad: str)->int:
    """Cuenta cuántas cadenas dentro de la lista contienen una subcadena dada (sin distinción de mayúsculas).

    Par:
        cadenas (list[str]): Lista de cadenas de texto donde buscar.
        subcad (str): Subcadena a buscar dentro de cada elemento.

    Devuelve:
        int: Número de cadenas que contienen la subcadena especificada.
    """
    cont = 0
    for cad in cadenas:
        if subcad.lower() in cad.lower():
            cont += 1
    return cont

def longitud_max_min(cadenas: list[str])-> None:
    """Encuentra e imprime la cadena de mayor y menor longitud dentro de la lista.

    Par:
        cadenas (list[str]): Lista de cadenas de texto a analizar.

    Devuelve:
        None: Imprime el resultado en consola o muestra un mensaje si la lista está vacía.
    """
    if not cadenas:
        print("Error. La lista está vacía")
        return None
    maximo = cadenas[0]
    minimo = cadenas[0]
    for subcad in cadenas:
        if len(subcad) > len(maximo):
            maximo = subcad
        elif len(subcad) < len(minimo):
            minimo =subcad

    print(f"Cadena más larga: {maximo}\n")
    print(f"Cadena más corta: {minimo}\n")

def mismo_objeto(cadenas: list[str], ind1: int, ind2: int)->bool | None:
    """Comprueba mediante el operador 'is' si dos elementos de la lista apuntan a la misma dirección de memoria.

    Par:
        cadenas (list[str]): Lista de cadenas de texto.
        ind1 (int): Índice del primer elemento a comparar.
        ind2 (int): Índice del segundo elemento a comparar.

    Devuelve:
        bool | None: True si son el mismo objeto, False si son objetos distintos, 
                     o None si la lista está vacía o un índice está fuera de rango.
    """
    if not cadenas:
        print("Error. La lista está vacía")
        return None
    
    
        
    if not (0 <= ind1 < len(cadenas) and 0 <= ind2 < len(cadenas)):
        print("Error. Uno o ambos índices están fuera de rango.")
        return

    if cadenas[ind1] is cadenas[ind2]:
        return True
    else:
        return False

def mayor_diez(cadenas: list[str])->bool | None:
    """Verifica si al menos una de las cadenas de la lista supera los 10 caracteres de longitud.

    Par:
        cadenas (list[str]): Lista de cadenas de texto a verificar.

    Devuelve:
        bool | None: True si existe al menos una cadena con más de 10 caracteres, 
                     False en caso contrario, o None si la lista está vacía.
    """
    if not cadenas:
        print("Error. La lista está vacía")
        return None

    for cad in cadenas:
        if len(cad)>10:
            return True
    else:
        return False

def main():
    cadenas = pedir_cadenas()
    opcion = ""
    while opcion != "6":
        opcion = mostrar_menu()

        match opcion:
            case "1":
                letra = input("Introduzca la letra por la que quiere que comiencen las cadenas: ")
                imprimir_por_inicial(cadenas, letra)

            case "2":
                subcad = input("Introduzca la subcadena a contabilizar: ")
                rep = contar_subcadenas(cadenas, subcad)
                print(f"Número de repeticiones de {subcad}: {rep}")

            case "3":
                longitud_max_min(cadenas)

            case "4":
                idx1 = int(input("Introduce el primer índice a comparar: "))
                idx2 = int(input("Introduce el segundo índice a comparar: "))
                if mismo_objeto(cadenas, idx1, idx2):
                    print("Las cadenas son el mismo objeto")
                else:
                    print("Las cadenas no son el mismo objeto")

            case "5":
                if mayor_diez(cadenas):
                    print("Al menos una cadena tiene longitud mayor que 10")
                else:
                    print("Ninguna cadena tiene longitud mayor que 10")

            case "6":
                print("Fin del programa")

if __name__ == "__main__":
    main()
```

```python
#Ejercicio 11
import re
from datetime import datetime

#Patrón dd/mm/aaaa

patron_fecha = re.compile(r"\b(\d{1,2})/(\d{1,2})/(\d{4})\b")



def informe_anonimo(texto: str)->str:
    """Anonimiza un informe médico simulado eliminando datos sensibles.

    Sustituye la fecha de nacimiento por la edad del paciente en el momento de
    la consulta, el resto de las fechas por la palabra "FECHA", el nombre del
    paciente por "PACIENTE" y el nombre del médico por "MEDICO".

    Args:
        texto (str): Cadena de texto que contiene el informe clínico completo
          en formato multilínea.

    Returns:
        str: Devuelve el informe clínico anonimizado con las sustituciones
        aplicadas.
    """
    match_nacimiento = re.search(r"Fecha de nacimiento:\s*(\d{1,2}/\d{1,2}/\d{4})", texto)
    fecha_nacimiento = match_nacimiento.group(1)
    fechas = patron_fecha.findall(texto)
    fecha_cons = f"{fechas[1][0]}/{fechas[1][1]}/{fechas[1][2]}" #nos qeudamos con la segunda fecha encontrada
    fn = datetime.strptime(fecha_nacimiento, "%d/%m/%Y")
    fc = datetime.strptime(fecha_cons, "%d/%m/%Y")

    if (fc.month, fc.day) < (fn.month, fn.day): #si no ha cumplido años todavía
        edad = fc.year -fn.year -1
    else:
        edad = fc.year - fn.year

    #Sustituimos en el texto
    texto = re.sub(
        r"Fecha de nacimiento:\s*\d{1,2}/\d{1,2}/\d{4}",
        f"Edad: {edad}",
        texto
    )

    texto = patron_fecha.sub("FECHA", texto)
    texto = re.sub(r"(Nombre:\s*).*", r"\1PACIENTE", texto)
    texto = re.sub(r"Dr[a]?\.\s+[\wÁÉÍÓÚáéíóúÑñ]+", "MEDICO", texto)

    return texto

def main():
    informe = """Informe clínico de Urgencias
Nombre: Juan Pérez López
Género: Hombre
Fecha de nacimiento: 12/05/1980
Motivo de consulta: dolor abdominal
12/05/2023 Dr. Ramírez
Tratamiento: reposo y analgésicos"""
    resultado = informe_anonimo(informe)
    print(resultado)

if __name__ == "__main__":
    main()
```

```python
#Ejercicio 12
def is_anagram(word1:str, word2:str)->bool:
    """Determina si dos cadenas de texto son anagramas.

    Ignora diferencias entre mayúsculas y minúsculas, así como los espacios en
    blanco dentro de las cadenas.

    Args:
        word1 (str): Primera cadena de texto a comparar.
        word2 (str): Segunda cadena de texto a comparar.

    Returns:
        bool: Devuelve True si las dos cadenas son anagramas, o False en caso
        contrario.
    """
    pal1 = word1.lower().replace(" ", "")#Convertimos a minúsculas y elimnamos espacios
    pal2 = word2.lower().replace(" ", "")#Convertimos a minúsculas y elimnamos espacios

    ordenado1 = "".join(sorted(pal1))
    ordenado2 = "".join(sorted(pal2))

    if ordenado1 == ordenado2:
        return True
    else:
        return False

def group_anagrams(words: list[str])->list[list]:
    """Agrupa una lista de palabras según sean anagramas entre sí.

    Mantiene el orden de primera aparición tanto para la creación de los
    grupos como para los elementos dentro de cada grupo. Si una palabra no
    tiene anagramas, forma un grupo individual. Reutiliza la función
    is_anagram.

    Args:
        words (list[str]): Lista de palabras o cadenas de texto a agrupar.

    Returns:
        list[list[str]]: Devuelve una lista de sublistas, donde cada sublista
        contiene las palabras que son anagramas entre sí.
    """
    groups = []
    for word in words:
        in_group = False
        for group in groups: #si la palabra pertenece a un groupo previo
            if is_anagram(word, group[0]):
                group.append(word)
                in_group = True
                break
        if not in_group:
            groups.append([word])
    return groups

def largest_group(groups: list[list])->list[str]:
    """Devuelve el grupo con mayor número de palabras a partir de una lista de grupos.

    En caso de empate en la cantidad de elementos entre dos o más grupos,
    mantiene y devuelve el primero que haya aparecido. Si la lista recibida
    está vacía, devuelve una lista vacía.

    Args:
        groups (list[list]): Lista de sublistas de palabras (resultado de
          group_anagrams).

    Returns:
        list[str]: Devuelve la sublista que contiene el grupo más numeroso de
        palabras.
    """
    mayor = []
    if not groups:
        return mayor
    for group in groups:
        if len(group)>len(mayor):
            mayor = group
    return mayor

def find_anagrams(word:str, words: list[str])->list[str]:
    """Encuentra los anagramas de una palabra dentro de una lista de palabras.

    Excluye cualquier palabra que sea exactamente igual a la palabra dada
    (ignorando mayúsculas y minúsculas). Si un anagrama válido aparece varias
    veces en la lista, se incluye en el resultado tantas veces como aparezca.

    Args:
        word (str): Palabra objetivo para buscar sus anagramas.
        words (list[str]): Lista de palabras candidates en las que buscar.

    Returns:
        list[str]: Devuelve la lista con todas las apariciones de los anagramas
        válidos encontrados.
    """
    anagrams = []
    for element in words:
        if element.lower() != word.lower():
            if is_anagram(word, element):
                anagrams.append(element)
    return anagrams

def main():
    print("--- 1. Pruebas de is_anagram ---")
    print("¿'Roma' y 'amor' son anagramas?:", is_anagram("Roma", "amor"))
    print(
        "¿'Tom Marvolo Riddle' e 'I am Lord Voldemort' son anagramas?:",
        is_anagram("Tom Marvolo Riddle", "I am Lord Voldemort"),
    )
    print("¿'Hola' y 'Mundo' son anagramas?:", is_anagram("Hola", "Mundo"))

    print("\n--- 2. Pruebas de group_anagrams ---")
    palabras_ejemplo = [
        "Roma",
        "amor",
        "Ramo",
        "hola",
        "aloh",
        "Mundo",
        "oMar",
        "mora",
    ]
    grupos = group_anagrams(palabras_ejemplo)
    print("Lista original:", palabras_ejemplo)
    print("Grupos formados:", grupos)

    print("\n--- 3. Pruebas de largest_group ---")
    grupo_mas_grande = largest_group(grupos)
    print("El grupo más grande es:", grupo_mas_grande)
    print("Prueba con lista vacía:", largest_group([]))

    print("\n--- 4. Pruebas de find_anagrams ---")
    palabra_busqueda = "Roma"
    lista_busqueda = [
        "roma",
        "Roma",
        "amor",
        "amor",
        "ramo",
        "perro",
        "mora",
        "MUNDO",
    ]
    anagramas_encontrados = find_anagrams(palabra_busqueda, lista_busqueda)
    print(f"Buscando anagramas para '{palabra_busqueda}' en:", lista_busqueda)
    print("Anagramas encontrados:", anagramas_encontrados)


if __name__ == "__main__":
    main()
```
